For this research question, I plan to use K-nearest neighbors (KNN) classification to predict whether a team receives an automatic or at-large NCAA tournament bid based on its regional wins and losses. This method is appropriate because we are predicting a categorical outcome using two numerical predictors.

One limitation of KNN is that its performance depends on the number of neighbors (K) chosen. A K value that is too small may lead to overfitting, while a value that is too large may lead to underfitting. The model may also struggle if teams receiving different bid types have similar regional records.

To prepare the data, I will first select the relevant columns, remove missing values, and convert the bid variable into a factor. I will then split the data into 75% training and 25% testing, using stratified sampling to maintain similar proportions of each bid type. I will standardize the numerical predictors because KNN uses distances to make predictions. Next, I will use 5-fold cross-validation on the training set to test different K values and select the one with the highest average accuracy. This means I won't need a separate validation set.

Finally, I will evaluate the model on the testing set using accuracy and a confusion matrix. Accuracy will show how often the model correctly predicts the bid type, while the confusion matrix will help identify which categories the model has difficulty predicting.
